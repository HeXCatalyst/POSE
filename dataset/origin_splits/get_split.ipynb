```python
test_list_path = "./test_list.txt"
test_out_list_path = "./test_out_list.txt"
train_list_path = "./train_list.txt"
```

```python
test_list = []
train_list = []
test_out_list = []
idx = 0 # 保证大家共用唯一的下标
with open(test_list_path, "r") as f:
    for line in f:
        test_list.append([idx]+line.strip().split("\t")) # 划分为图片地址和种子
        idx += 1
with open(train_list_path, "r") as f:
    for line in f:
        train_list.append([idx] + line.strip().split("\t"))
        idx += 1
with open(test_out_list_path, "r") as f:
    for line in f:
        test_out_list.append([idx] + line.strip().split("\t"))
        idx += 1
```

```python
from tqdm import tqdm
st = set()
for lst in tqdm(test_out_list):
    name = lst[1].split("/")[-2]
    st.add(name)
    
```
